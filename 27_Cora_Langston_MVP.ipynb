# ⏰ Cora Langston — The Time Traveler | Voice Aging MVP

Character: **Cora Langston** — a scientist who travels time, accumulating centuries of experience across a single lifetime.

This notebook demonstrates the voice aging process across four stages of her life, plus an emotional range test, using the Qwen3-TTS Voice Design model.

```python
!pip install -q soundfile
# Assuming qwen-tts is already installed or available in the environment
```

```python
import os
import gc
import torch
import numpy as np
import soundfile as sf
from IPython.display import Audio, display
from qwen_tts import Qwen3TTSModel

# Initialize model
model_id = "Qwen/Qwen3-TTS-12Hz-1.7B-VoiceDesign"
model = Qwen3TTSModel.from_pretrained(
    model_id, 
    device_map="cuda:0", 
    dtype=torch.bfloat16, 
    attn_implementation="sdpa"
)
print("Model loaded successfully!")
```

```python
os.makedirs("outputs", exist_ok=True)

def generate_and_play(text, instruct, filename):
    print(f"Generating: {filename}")
    print(f"Text: {text}")
    print(f"Instruct: {instruct}\n")
    
    wavs, sr = model.generate_voice_design(text, "English", instruct)
    
    sf.write(filename, wavs[0], sr)
    display(Audio(filename))
    
    # Cleanup VRAM
    del wavs
    gc.collect()
    torch.cuda.empty_cache()
```

## ⏰ Youth (22) — The Explorer

**Core Personality:** an endlessly curious, slightly breathless female scientist — speaks quickly when excited, slows to profound stillness when serious, warmth for all living things, carries impossible knowledge lightly

**Voice Modifier:** bright and breathless, words tumbling over each other with excitement, wide-eyed wonder, quick and elastic voice

```python
personality_core = "an endlessly curious, slightly breathless female scientist — speaks quickly when excited, slows to profound stillness when serious, warmth for all living things, carries impossible knowledge lightly"
voice_modifier = "bright and breathless, words tumbling over each other with excitement, wide-eyed wonder, quick and elastic voice"
instruct = f"{personality_core}, {voice_modifier}"

lines = [
    "Do you understand what this means?! We can go anywhere. Anywhen. The entirety of history is just — it's just sitting there, waiting!",
    "I know the theory says it's impossible. The theory was written by people who never tried it. I tried it. It's very possible.",
    "I've already seen next Tuesday. It's fine, by the way. Nothing to worry about. For you, anyway."
]

for i, line in enumerate(lines):
    generate_and_play(line, instruct, f"outputs/cora_youth_line_{i+1}.wav")
```

## ⏰ Prime (36) — The Veteran

**Voice Modifier:** still warm and curious but with infinite layers of knowledge underneath, the excitement more controlled, slightly otherworldly cadence from too many timelines

```python
voice_modifier = "still warm and curious but with infinite layers of knowledge underneath, the excitement more controlled, slightly otherworldly cadence from too many timelines"
instruct = f"{personality_core}, {voice_modifier}"

lines = [
    "I've watched this moment happen seventeen different ways. In fourteen of them, we make it. I find that — on balance — encouraging.",
    "Time isn't a river. It's more like — a garden. Full of paths. Some overgrown. Some not yet made. All of them real.",
    "I've met you before. Several times, actually. You always say the same thing at this point. It's rather sweet."
]

for i, line in enumerate(lines):
    generate_and_play(line, instruct, f"outputs/cora_prime_line_{i+1}.wav")
```

## ⏰ Middle (51) — The Weighted

**Voice Modifier:** slower and more deliberate, carrying centuries of witnessed events, profound and a little sad, still warm but the wonder is now tinged with loss

```python
voice_modifier = "slower and more deliberate, carrying centuries of witnessed events, profound and a little sad, still warm but the wonder is now tinged with loss"
instruct = f"{personality_core}, {voice_modifier}"

lines = [
    "I have watched civilizations begin. I have watched them end. I've learned not to pick favorites.",
    "The hardest part isn't the paradoxes. It's watching people make mistakes you know are coming and understanding why you cannot warn them.",
    "I stopped keeping a diary somewhere around my four hundredth year of subjective experience. There wasn't enough paper in any timeline."
]

for i, line in enumerate(lines):
    generate_and_play(line, instruct, f"outputs/cora_middle_line_{i+1}.wav")
```

## ⏰ Elder (68) — The Serene

**Voice Modifier:** deeply calm, vast patience, speaks as someone who has made peace with the entire span of existence, occasional wry humor from impossible perspective

```python
voice_modifier = "deeply calm, vast patience, speaks as someone who has made peace with the entire span of existence, occasional wry humor from impossible perspective"
instruct = f"{personality_core}, {voice_modifier}"

lines = [
    "I've been asked what the future holds. The future holds everything. That's rather the point.",
    "I am technically 68 years old. I am experientially somewhere between two thousand and the concept of Tuesday. I've stopped reconciling these figures.",
    "Every moment that has ever happened is still happening somewhere in the structure of time. I find that — on my better days — genuinely beautiful."
]

for i, line in enumerate(lines):
    generate_and_play(line, instruct, f"outputs/cora_elder_line_{i+1}.wav")
```

## 🎭 Emotional Range Test

Testing the same line in three different emotional registers.

```python
target_line = "I've watched this moment happen seventeen different ways. In fourteen of them, we make it."

emotions = {
    "breathless excitement": f"{personality_core}, speaking with breathless excitement, eager and energetic",
    "calm certainty": f"{personality_core}, speaking with calm certainty, steady and reassuring",
    "quiet grief": f"{personality_core}, speaking with quiet grief, sorrowful and mourning"
}

for emotion_name, instruct in emotions.items():
    filename = f"outputs/cora_emotion_{emotion_name.replace(' ', '_')}.wav"
    generate_and_play(target_line, instruct, filename)
```
