# 🌌 Dorian Slate — The Reluctant Prophet | Voice Aging MVP

Character: **Dorian Slate** — ordinary man who discovers he's prophesied to save everything, and never fully reconciles with it.

Using Alibaba's Qwen3-TTS Voice Design model to simulate voice aging across 4 life stages.

```python
# @title ⚙️ Setup and Installation
!pip install -q qwen-tts accelerate soundfile

import torch
import gc
import os
import numpy as np
import soundfile as sf
from IPython.display import Audio, display
from qwen_tts import Qwen3TTSModel

def cleanup():
    gc.collect()
    torch.cuda.empty_cache()
    
os.makedirs('outputs', exist_ok=True)
```

```python
# @title 🧠 Load Model
model_id = "Qwen/Qwen3-TTS-12Hz-1.7B-VoiceDesign"

print(f"Loading {model_id}...")
model = Qwen3TTSModel.from_pretrained(
    model_id, 
    device_map="cuda:0", 
    dtype=torch.bfloat16, 
    attn_implementation="sdpa"
)
print("Model loaded successfully!")
cleanup()
```

```python
# @title 🗣️ Generation Utilities
def generate_and_play(text, instruct, filename):
    print(f"Generating: {filename}")
    print(f"Text: '{text}'")
    print(f"Instruct: '{instruct}'")
    
    wavs, sr = model.generate_voice_design(
        text=text, 
        language="English", 
        instruct=instruct
    )
    
    filepath = f"outputs/{filename}.wav"
    sf.write(filepath, wavs[0], sr)
    
    display(Audio(filepath))
    print("-" * 40)
    cleanup()

personality_core = "a quiet, ordinary male — plain speech patterns, no rhetorical flourish, which makes the weight of his words more striking; the grandness of his destiny sits wrong on his shoulders and it shows"
```

## Stage 1: Youth (20) — The Ordinary

```python
voice_modifier = "genuinely plain and unassuming, slightly confused, a normal person in an abnormal situation, not performing normalcy — actually normal"
instruct = f"{personality_core}, {voice_modifier}"

lines = [
    "I'm not the person you're looking for. I want to be very clear about that. I'm a nobody from a very small town.",
    "They keep saying I'm chosen. I've looked into it. The original prophecy is extremely vague. It could be describing several people.",
    "I'll help. Of course I'll help. But I need someone to explain — very slowly — why it has to be me."
]

for i, line in enumerate(lines, 1):
    generate_and_play(line, instruct, f"Dorian_01_Youth_Line{i}")
```

## Stage 2: Prime (33) — The Reluctant

```python
voice_modifier = "still plain speech but carrying impossible weight, quiet resolve that doesn't sit comfortably on him, pauses that hold entire histories"
instruct = f"{personality_core}, {voice_modifier}"

lines = [
    "I've done the things the prophecy required. I didn't enjoy most of them. I'd do them again. That doesn't make me a hero. It makes me someone who ran out of options.",
    "People want me to be inspiring. I'm not inspiring. I'm just stubborn. And I have very specific information about how this ends.",
    "I think about the version of me that stayed home. I hope he's happy. I genuinely do."
]

for i, line in enumerate(lines, 1):
    generate_and_play(line, instruct, f"Dorian_02_Prime_Line{i}")
```

## Stage 3: Middle (48) — The Carrier

```python
voice_modifier = "slower and heavier, the plain speech now profound through context, someone who has carried something immense for decades and can no longer pretend it's light"
instruct = f"{personality_core}, {voice_modifier}"

lines = [
    "I've been called a savior for twenty-eight years. I've never found a way to respond to that that didn't feel dishonest.",
    "The prophecy was right about everything except how it would feel. No one thinks to put that part in prophecies.",
    "I used to want my ordinary life back. Now I'm not sure what that means anymore. The wanting is still there. The life isn't."
]

for i, line in enumerate(lines, 1):
    generate_and_play(line, instruct, f"Dorian_03_Middle_Line{i}")
```

## Stage 4: Elder (67) — The Transcendent

```python
voice_modifier = "deep and unhurried, the plain speech now carrying something almost cosmic in its simplicity, peace made with impossibility, gentle and vast"
instruct = f"{personality_core}, {voice_modifier}"

lines = [
    "I've accepted that this is what I am. It took most of my life. But I got there.",
    "The prophecy ended. The world didn't end with it. People keep living. That was always the point.",
    "If someone tells you you're the chosen one, believe them. Argue about it later. Do the thing first. The arguing can wait."
]

for i, line in enumerate(lines, 1):
    generate_and_play(line, instruct, f"Dorian_04_Elder_Line{i}")
```

## 🎭 Emotional Range Test

```python
line = "I'm not the person you're looking for. I want to be very clear about that."

emotions = [
    "confused and genuine",
    "resigned and tired",
    "quietly certain"
]

for i, emotion in enumerate(emotions, 1):
    instruct = f"{personality_core}, speaking in a way that is {emotion}"
    generate_and_play(line, instruct, f"Dorian_Emotion_Test_0{i}")
```
