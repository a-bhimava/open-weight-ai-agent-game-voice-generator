```python
import os
import gc
import torch
import numpy as np
import soundfile as sf
from IPython.display import Audio, display
from qwen_tts import Qwen3TTSModel

# Output directory
out_dir = "nora_outputs"
os.makedirs(out_dir, exist_ok=True)

# Load Model
model_id = "Qwen/Qwen3-TTS-12Hz-1.7B-VoiceDesign"
print(f"Loading {model_id} on T4 GPU...")
model = Qwen3TTSModel.from_pretrained(
    model_id,
    device_map="cuda:0",
    dtype=torch.bfloat16,
    attn_implementation="sdpa"
)
print("Model loaded successfully!")

def generate_and_play(text, instruct, filename):
    print(f"\nGenerating: {filename}")
    print(f"Prompt: {instruct}")
    wavs, sr = model.generate_voice_design(text, "English", instruct)
    sf.write(filename, wavs[0], sr)
    display(Audio(filename))
    
    # Cleanup VRAM
    gc.collect()
    torch.cuda.empty_cache()

personality_core = "an intensely quiet, precise female hacker — speaks rarely, every word chosen with care, flat delivery that occasionally cracks to reveal deep moral fury, never raises voice"
```

## Youth (20) — The Recluse

```python
voice_modifier = "quiet and flat, slightly awkward socially, immense intelligence evident in precise word choice, the rage barely a simmer"
instruct = f"{personality_core} {voice_modifier}"

lines = [
    "Most people don't understand how the world actually works. I do. I'm not sure which of us is luckier.",
    "I don't need anyone to believe me. The data speaks for itself. It always does, if you know how to listen.",
    "I'm not angry. I'm just paying very close attention. There's a difference."
]

for i, line in enumerate(lines, 1):
    generate_and_play(line, instruct, f"{out_dir}/youth_{i}.wav")
```

## Prime (32) — The Operative

```python
voice_modifier = "controlled and precise, dangerous stillness, the flat delivery now clearly intentional — a weapon, not a limitation"
instruct = f"{personality_core} {voice_modifier}"

lines = [
    "I've been inside every major government database on three continents. Not because I wanted to. Because I needed to know if they could be trusted.",
    "I don't break systems. I show people what the systems already are. The breaking was done long before I arrived.",
    "If you're looking for someone to make you feel better about this situation, I am genuinely the wrong person to call."
]

for i, line in enumerate(lines, 1):
    generate_and_play(line, instruct, f"{out_dir}/prime_{i}.wav")
```

## Middle (47) — The Ghost

```python
voice_modifier = "even quieter, the pauses longer, someone who has spent years speaking to almost no one, still precise but with a new weariness"
instruct = f"{personality_core} {voice_modifier}"

lines = [
    "I deleted myself from seven national registries. I did it gradually, over four years. Nobody noticed. That told me everything I needed to know.",
    "The people I exposed are still in power. Most of them. That used to make me angry. Now it just makes me strategic.",
    "I live very quietly now. I watch. I wait. And occasionally, I remind certain very powerful people that I'm still watching."
]

for i, line in enumerate(lines, 1):
    generate_and_play(line, instruct, f"{out_dir}/middle_{i}.wav")
```

## Elder (64) — The Vanished

```python
voice_modifier = "barely above a murmur, serene and final, someone who has made complete peace with invisibility, faint warmth surfacing"
instruct = f"{personality_core} {voice_modifier}"

lines = [
    "No one knows my name. No one knows my face. No one knows what I've done. I find that — on the whole — to be the correct outcome.",
    "I spent forty years fighting with information as my only weapon. Information is infinite. That's the one advantage I always had.",
    "I'm not gone. I just got very, very good at being somewhere else."
]

for i, line in enumerate(lines, 1):
    generate_and_play(line, instruct, f"{out_dir}/elder_{i}.wav")
```

## 🎭 Emotional Range Test

```python
line = "I don't break systems. I show people what the systems already are."
emotions = [
    ("flat_factual", "[flat and factual]"),
    ("barely_contained_fury", "[barely contained fury]"),
    ("quiet_resignation", "[quiet resignation]")
]

for emotion_name, modifier in emotions:
    instruct = f"{personality_core} {modifier}"
    generate_and_play(line, instruct, f"{out_dir}/emotion_{emotion_name}.wav")
```
