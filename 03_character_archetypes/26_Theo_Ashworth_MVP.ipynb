# 👑 Theo Ashworth — The Fallen King | Voice Aging MVP

Character: **Theo Ashworth** — a beloved noble ruler stripped of everything, who becomes a wise wanderer.

This notebook demonstrates the voice aging process across four stages of his life using the Qwen3-TTS Voice Design model.

```python
!pip install -q soundfile
# Assuming qwen-tts is already installed or available in the environment
```

```python
import os
import gc
import torch
import numpy as np
import soundfile as sf
from IPython.display import Audio, display
from qwen_tts import Qwen3TTSModel

# Initialize model
model_id = "Qwen/Qwen3-TTS-12Hz-1.7B-VoiceDesign"
model = Qwen3TTSModel.from_pretrained(
    model_id, 
    device_map="cuda:0", 
    dtype=torch.bfloat16, 
    attn_implementation="sdpa"
)
print("Model loaded successfully!")
```

```python
os.makedirs("outputs", exist_ok=True)

def generate_and_play(text, instruct, filename):
    print(f"Generating: {filename}")
    print(f"Text: {text}")
    print(f"Instruct: {instruct}\n")
    
    wavs, sr = model.generate_voice_design(text, "English", instruct)
    
    sf.write(filename, wavs[0], sr)
    display(Audio(filename))
    
    # Cleanup VRAM
    del wavs
    gc.collect()
    torch.cuda.empty_cache()
```

## 👑 Youth (24) — The Crowned

**Core Personality:** a noble, resonant, deeply principled male ruler — formal and measured speech, immense dignity even in defeat, voice that was built for proclamations but learned humility

**Voice Modifier:** ceremonial nobility, formal cadence, young king's earnest conviction, strong and clear with untested idealism

```python
personality_core = "a noble, resonant, deeply principled male ruler — formal and measured speech, immense dignity even in defeat, voice that was built for proclamations but learned humility"
voice_modifier = "ceremonial nobility, formal cadence, young king's earnest conviction, strong and clear with untested idealism"
instruct = f"{personality_core}, {voice_modifier}"

lines = [
    "I accept this crown not as a prize but as a responsibility. Every life in this kingdom is mine to answer for. I do not take that lightly.",
    "A king who rules through fear builds walls. A king who rules through trust builds a nation. I intend to build a nation.",
    "I have been prepared for this role my entire life. I am still not certain anyone can truly be prepared for it."
]

for i, line in enumerate(lines):
    generate_and_play(line, instruct, f"outputs/theo_youth_line_{i+1}.wav")
```

## 👑 Prime (39) — The Ruler

**Voice Modifier:** deep commanding authority, formal but warmer, experience tempering the ceremony, decisions made in his voice's weight

```python
voice_modifier = "deep commanding authority, formal but warmer, experience tempering the ceremony, decisions made in his voice's weight"
instruct = f"{personality_core}, {voice_modifier}"

lines = [
    "Governing is not what they teach you it is. It is thousands of small decisions that compound into a century.",
    "I have made enemies by trying to be fair. That tells me I am doing something correctly.",
    "The crown is heavier than it looks. That is not a metaphor. It is the literal truth, and also a metaphor."
]

for i, line in enumerate(lines):
    generate_and_play(line, instruct, f"outputs/theo_prime_line_{i+1}.wav")
```

## 👑 Middle (54) — The Fallen

**Voice Modifier:** stripped of ceremony, raw and broken, the nobility still present but wounded, someone rebuilding from nothing

```python
voice_modifier = "stripped of ceremony, raw and broken, the nobility still present but wounded, someone rebuilding from nothing"
instruct = f"{personality_core}, {voice_modifier}"

lines = [
    "I had a kingdom. I had a purpose. I had thirty years of decisions I believed in. All of it taken in a single night.",
    "What they couldn't take was what I learned. Every face. Every town. Every choice. That belongs only to me.",
    "I am more valuable now, stripped of the crown, than I ever was wearing it. That is the cruelest lesson leadership teaches."
]

for i, line in enumerate(lines):
    generate_and_play(line, instruct, f"outputs/theo_middle_line_{i+1}.wav")
```

## 👑 Elder (71) — The Wanderer

**Voice Modifier:** quiet dignity, unhurried, the formality distilled into something simpler and truer, wisdom without the performance

```python
voice_modifier = "quiet dignity, unhurried, the formality distilled into something simpler and truer, wisdom without the performance"
instruct = f"{personality_core}, {voice_modifier}"

lines = [
    "I governed for thirty years and wandered for twenty. The wandering taught me more.",
    "Power is temporary. What you do with it is not. I try to remember that more kindly now than I did when I had it.",
    "A king without a kingdom is just a man with an unusually long memory and a great deal of regret. I've made my peace with that."
]

for i, line in enumerate(lines):
    generate_and_play(line, instruct, f"outputs/theo_elder_line_{i+1}.wav")
```
