# 🩺 Lena Stavros — The Battle Medic | Voice Aging MVP

Character: **Lena Stavros** — idealistic doctor who ends up healing people in impossible circumstances.

Using Alibaba's Qwen3-TTS Voice Design model to simulate voice aging across 4 life stages.

```python
# @title ⚙️ Setup and Installation
!pip install -q qwen-tts accelerate soundfile

import torch
import gc
import os
import numpy as np
import soundfile as sf
from IPython.display import Audio, display
from qwen_tts import Qwen3TTSModel

def cleanup():
    gc.collect()
    torch.cuda.empty_cache()
    
os.makedirs('outputs', exist_ok=True)
```

```python
# @title 🧠 Load Model
model_id = "Qwen/Qwen3-TTS-12Hz-1.7B-VoiceDesign"

print(f"Loading {model_id}...")
model = Qwen3TTSModel.from_pretrained(
    model_id, 
    device_map="cuda:0", 
    dtype=torch.bfloat16, 
    attn_implementation="sdpa"
)
print("Model loaded successfully!")
cleanup()
```

```python
# @title 🗣️ Generation Utilities
def generate_and_play(text, instruct, filename):
    print(f"Generating: {filename}")
    print(f"Text: '{text}'")
    print(f"Instruct: '{instruct}'")
    
    wavs, sr = model.generate_voice_design(
        text=text, 
        language="English", 
        instruct=instruct
    )
    
    filepath = f"outputs/{filename}.wav"
    sf.write(filepath, wavs[0], sr)
    
    display(Audio(filepath))
    print("-" * 40)
    cleanup()

personality_core = "a fierce, pragmatic female doctor — no-nonsense and direct, deep compassion expressed through action not sentiment, voice that stays steady under pressure, becomes rougher and more tired with age"
```

## Stage 1: Youth (25) — The Idealist

```python
voice_modifier = "clear and earnest, medical precision in word choice, genuine idealism, the toughness is there but not yet fully formed"
instruct = f"{personality_core}, {voice_modifier}"

lines = [
    "People ask why I chose trauma surgery. Because someone has to be there at the worst moment of someone else's life, and actually help.",
    "I don't do this for glory. The OR is not a place for glory. It's a place for focus and skill and not making mistakes.",
    "Every patient gets everything I have. That's not noble. That's just the job. If you're not willing to do the job, get out of the way."
]

for i, line in enumerate(lines, 1):
    generate_and_play(line, instruct, f"Lena_01_Youth_Line{i}")
```

## Stage 2: Prime (38) — The Surgeon

```python
voice_modifier = "professional calm under enormous pressure, controlled and precise, no wasted words, genuine authority in every syllable"
instruct = f"{personality_core}, {voice_modifier}"

lines = [
    "I've worked sixty-hour shifts in field hospitals with no electricity. After that, a difficult surgery in a proper OR feels like a vacation.",
    "The rule is simple: you are calm. You are focused. You do not panic. Whatever your patient is feeling, you give them your steadiness as a gift.",
    "I've lost patients I shouldn't have and saved ones I didn't expect to. Medicine humbles you constantly. That's the whole curriculum, really."
]

for i, line in enumerate(lines, 1):
    generate_and_play(line, instruct, f"Lena_02_Prime_Line{i}")
```

## Stage 3: Middle (52) — The Worn

```python
voice_modifier = "rougher and slower, the toughness carrying real weight now, weariness mixed with undiminished competence, voice of someone who has seen too much and kept working anyway"
instruct = f"{personality_core}, {voice_modifier}"

lines = [
    "Twenty-seven years. I've stopped counting the surgeries but I remember every patient I lost. That's not something the training prepares you for.",
    "People think doctors become detached. We don't. We just learn to carry it differently. The weight doesn't go away. You just get stronger.",
    "I train young surgeons now. The first thing I tell them: the day this stops being hard is the day you should quit."
]

for i, line in enumerate(lines, 1):
    generate_and_play(line, instruct, f"Lena_03_Middle_Line{i}")
```

## Stage 4: Elder (67) — The Veteran

```python
voice_modifier = "gravelly and slow, exhaustion finally audible, deep compassion fully integrated into every word, the pragmatism softened into something like grace"
instruct = f"{personality_core}, {voice_modifier}"

lines = [
    "I've spent forty years keeping people alive long enough to live the rest of their lives. I can't think of a better way to have spent a life.",
    "My hands aren't steady enough for the OR anymore. I find I'm somewhat at peace with that. Somewhat.",
    "Every patient was someone's everything. I tried to remember that. Every time. In the end, it's the only thing that mattered."
]

for i, line in enumerate(lines, 1):
    generate_and_play(line, instruct, f"Lena_04_Elder_Line{i}")
```
