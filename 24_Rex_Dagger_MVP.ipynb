```python
import os
import gc
import torch
import numpy as np
import soundfile as sf
from IPython.display import Audio, display
from qwen_tts import Qwen3TTSModel

# Output directory
out_dir = "rex_outputs"
os.makedirs(out_dir, exist_ok=True)

# Load Model
model_id = "Qwen/Qwen3-TTS-12Hz-1.7B-VoiceDesign"
print(f"Loading {model_id} on T4 GPU...")
model = Qwen3TTSModel.from_pretrained(
    model_id,
    device_map="cuda:0",
    dtype=torch.bfloat16,
    attn_implementation="sdpa"
)
print("Model loaded successfully!")

def generate_and_play(text, instruct, filename):
    print(f"\nGenerating: {filename}")
    print(f"Prompt: {instruct}")
    wavs, sr = model.generate_voice_design(text, "English", instruct)
    sf.write(filename, wavs[0], sr)
    display(Audio(filename))
    
    # Cleanup VRAM
    gc.collect()
    torch.cuda.empty_cache()

personality_core = "a roguish, quick-witted male space mercenary — charming and irreverent, talks fast to distract from how much he actually cares, warm heart buried under layers of sarcasm"
```

## Youth (21) — The Reckless

```python
voice_modifier = "fast, cocky, charming rogue energy, plays everything for laughs, genuine heart bleeding through the bravado"
instruct = f"{personality_core} {voice_modifier}"

lines = [
    "I have a plan. It's not a good plan. But in my experience, the good plans are the ones that get you killed.",
    "I'm not a hero. Let's be really clear about that. I'm a businessman who keeps accidentally doing heroic things.",
    "Look, I've been in worse situations than this. Well — roughly this bad. Okay, exactly this bad. Once."
]

for i, line in enumerate(lines, 1):
    generate_and_play(line, instruct, f"{out_dir}/youth_{i}.wav")
```

## Prime (35) — The Reluctant Hero

```python
voice_modifier = "still irreverent but with genuine weight beneath, the wit is slower now and more earned, warmth less hidden"
instruct = f"{personality_core} {voice_modifier}"

lines = [
    "I've saved the galaxy. Twice. Nobody builds statues of people like me. That's fine. Statues are a maintenance nightmare anyway.",
    "There's a crew that depends on me now. I didn't ask for that. I definitely didn't want it. And I would do absolutely anything for them.",
    "I used to think caring about things was a tactical weakness. I was wrong. It turns out it's the only thing that makes any of this worth doing."
]

for i, line in enumerate(lines, 1):
    generate_and_play(line, instruct, f"{out_dir}/prime_{i}.wav")
```

## Middle (50) — The Scarred

```python
voice_modifier = "slower and rougher, the humor still present but quieter, losses evident in the pauses, genuine warmth now unguarded"
instruct = f"{personality_core} {voice_modifier}"

lines = [
    "I've lost ships, crew, and a truly impressive number of personal items to the vacuum of space. I grieve them all equally. The crew more.",
    "Someone once told me I had no principles. They were right at the time. I've been collecting them ever since, one catastrophic mistake at a time.",
    "The galaxy is enormous and mostly trying to kill you. I've found that helps you appreciate the parts that aren't."
]

for i, line in enumerate(lines, 1):
    generate_and_play(line, instruct, f"{out_dir}/middle_{i}.wav")
```

## Elder (69) — The Legend

```python
voice_modifier = "gravelly and unhurried, the wit finally fully integrated with the warmth, storyteller's cadence, comfortable in his own skin at last"
instruct = f"{personality_core} {voice_modifier}"

lines = [
    "People ask me how I've survived this long in this business. Honestly? I have absolutely no idea. But I've enjoyed finding out.",
    "The secret to a long life in space is simple: never fly alone, never trust a free meal, and never, ever let anyone know how much you care.",
    "I've been called a scoundrel, a pirate, a rogue, and a hero. The hero one still makes me uncomfortable. I'm working on it."
]

for i, line in enumerate(lines, 1):
    generate_and_play(line, instruct, f"{out_dir}/elder_{i}.wav")
```
